# EdgeAI Hybrid Teaching Approach – Desktop Vision Prototyping

This notebook represents **Phase 2 of the hybrid approach** used while OpenMV hardware
is unavailable.

## Goal
- Learn vision concepts (color detection, edges, simple ML-style features)
- Use **Jupyter + OpenCV** as a *stand-in* for OpenMV's image pipeline
- Write logic that will later map directly to OpenMV APIs

## Mapping to OpenMV
| Desktop (OpenCV) | OpenMV |
|------------------|--------|
| cv2.imread()    | sensor.snapshot() |
| NumPy array     | image.Image |
| cv2.findContours | img.find_blobs() |


```
# Setup
import cv2
import numpy as np
import matplotlib.pyplot as plt

print('OpenCV version:', cv2.__version__)
```

## Create a Synthetic Camera Frame
We generate a fake image to simulate a camera snapshot.
This avoids external files and works everywhere.

```
# Create synthetic image
img = np.zeros((240, 320, 3), dtype=np.uint8)

# Draw colored objects (simulated targets)
cv2.rectangle(img, (50, 50), (150, 150), (0, 255, 0), -1)  # green box
cv2.circle(img, (240, 120), 30, (0, 0, 255), -1)           # red circle

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Simulated Camera Frame')
plt.axis('off')
```

## Color Detection (Blob Detection Analog)
This mirrors OpenMV's `find_blobs()` logic.

```
# Convert to HSV
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# Green color range
lower_green = np.array([40, 50, 50])
upper_green = np.array([80, 255, 255])

mask = cv2.inRange(hsv, lower_green, upper_green)

# Find contours
contours, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

output = img.copy()
for cnt in contours:
    x, y, w, h = cv2.boundingRect(cnt)
    cv2.rectangle(output, (x, y), (x+w, y+h), (255, 0, 0), 2)

plt.imshow(cv2.cvtColor(output, cv2.COLOR_BGR2RGB))
plt.title('Detected Green Object')
plt.axis('off')
```

## Edge Detection
Edge detection is frequently used before feature extraction or ML inference.

```
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
edges = cv2.Canny(gray, 50, 150)

plt.imshow(edges, cmap='gray')
plt.title('Edge Detection')
plt.axis('off')
```
